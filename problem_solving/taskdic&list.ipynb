## question 1, add number

```python
my_set = {1,2,3,}
print(my_set)
my_set.add(4)
print(my_set)
```

## question 2, remove

```python
my_set = {1,2,3,4}
print(my_set)
my_set.remove(2)
print(my_set)
```

## question 3, union

```python
#using operater
set1 = {1,2,3}
set2 = {3,4,5}
s = set1 | set2
print(s)
```

```python
#using union
set1 = {1,2,3}
set2 = {3,4,5}
s = set1.union(set2)
print(s)
```

## question 3, intersection

```python
set1 = {1,2,3}
set2 = {2,3,4}
s = set1.intersection(set2)
print(s)
```

## question4, max-min

```python
numbers = {10,2,3,44,6}
maximum = max(numbers)
minimum = min(numbers)
print(f"maximum:",maximum,"minimum:",minimum)
```

## Dictonary

```python
my_dict = {"name" : "yogs"
           ,"age" : "26",
           "city":"ktm" }
print(my_dict)
```

```python
#using for loop
my_dict = {"name": "yogs",
           "age": "26",
           "city":"ktm"}
for dict in my_dict:
    print(dict ,my_dict[dict], sep=",")
```

## question 2 dict

```python
#usinf [] brackets,if the key doesn't exist, error occured 
person = {'name':'Bob','age':30,'city':'Los Angeles','job':'Doctor'}
print("Name:" ,person["name"])
print("Job:",person["job"])
```

```python
#using get(),if the key doesn't exist,  avoid the error 
person = {'name':'Bob','age':30,'city':'Los Angeles','job':'Doctor'}
print("Name:",person.get("name"))
print("Job:",person.get("job"))
```

## creating nested dict 

```python
people = {"person1":
        {
        "name": "yogs",
        "age": 26,
        "city": "ktm",
        "job": "Engineer"
         },
    "person2": 
    {
       "name": "khaling",
        "age": 35,
       "city": "bkt",
        "job": "Doctor"
    }
}
print("Details of person1:")
for key, value in people["person1"].items():
    print(f"{key}: {value}")
```
