# Float

## question 1, checking float no is +ve

```python
num = float(input("enter the number"))
if num >0:
    print(f"{num},is positive number")
elif num ==0:
    print(f"{num}, is number zero")
else:
    print(f"{num} is negative")
```

## question 2, checking float within 

```python
num = float(input("enter the number"))
if num >= 0.0 and num <=1.0:
    print("inclusive")
else:
    print("error")
```

## question 7, checking float is greater

```python
num = float(input("enter a number:"))
rounded = round(num, 1)
if rounded > 5.5:
    print(f"{rounded} is greater than 5.5")
else:
    print(f"{rounded} is not greater than 5.5")
```

## question 8 , checking  *multiply float is greater than 10
