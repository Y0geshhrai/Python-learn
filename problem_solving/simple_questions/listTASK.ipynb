# List

## question 1, check if list is empty

```python
#using not operator
l = []
if not l:
    print("the list is empty")
else:
    print("list is not empty")
```

```python
#using len()

if len(l) == 0:
    print("the list is empty")
else:
    print("the list is not empty")
```

## question 2, determine first and last are same

```python
l = [10,20,30,10]
if l[0]==l[-1]:
    print("first and last elements are same")
else:
    print("first and last are not same")
```

## question 3, checking list contains more than

```python
s = [1,2,3,4,5,6,7,8]
if len(s) > 5:
    print("Long list")
else:
    print("short list")
```

## question 4 , checking specific value

```python
l = [2,3,4,5,6,7,8]
if 7 in l:
    print("7 is exist in list")
else:
    print("7 is not in the list")
```

## question 5, sum of no's >100 

```python
l = [10,20,40,40,50,10]
total = 0
for element in l:
   total = total + element
print(f"Total={total}"," High sum")
```

## question 7, checking all the elements in list  are even

```python
#using all()
number = [2,4,6,8,10]
if all(num % 2 == 0 for num in number):
    print("all the elements are even")
else:
    print("not all elements are equal")
```

## question 8, sorted in ascending order

```python
number = [2,3,4,1,6,8,7]
if number == sorted(number):
    print("the list is sorted in ascending order")
else:
    print("the list is not sorted in ascending order")

```

## question 9, duplicate values 

```python
num = [1,2,3,4,4,3,5,6]
if len(num) != len(set(num)): #set(num) removes duplicate
    print("Has Duplicate")
else:
    print("Doesn't have duplicate")
```

## question 10,  contains only positive numbers

```python
#using for loop
l = [1,2,3,4,5,6,7]
positive = True
for num in l:
    if num <= 0:
        positive = False
        break
if positive:
    print("All positive")
else:
    print("Negative")
```

```python
#using all() function
number = [1,2,3,4,5]
if all(num>0 for num in number): # all() evalutes to True if satisfy condition
    print("All positive")
else:
    print("negative")
```
