# Dictionary

## question 1, checking key exist

```python
#using in operator
dict = {"name":"ram","age":"20","city":"KTM"}
if "name" in dict: # returns true if key is in it , otherwsie it returns false
    print("the key 'name' in the dict")
else:
    print("the key 'name' is not in dictionary")
```

```python
#using get()
dict = {"name":"ram","age":"10","city":"pkr"}
if dict.get("name"):
    print("the key 'name' is in the dictionary")
else:
    print("the key 'name' is not in the dictionary")
```

```python
#using in operator
student = {"name":"matt","subjects":"IT","marks":45}
if "marks" in student:
    if student["marks"] >= 40:
        print("pass")
    else:
        print("fail")
```

```python
#using get()
student = {"name":"matt","subjects":"IT","marks":42}
marks = student.get("marks")
#if marks is not None: using None which returns 
if marks >=40:
    print("pass")
else:
    print("fail")
```

```python
student = {'name': 'Alice', 'marks': [45, 50, 60, 70]}

if 'marks' in student:
    marks = student['marks']
    if isinstance(marks, list):
        if all(mark >= 40 for mark in marks):
            print("pass")
        else:
            print("fail: not all marks are >= 40")
    else:
        print("fail: 'marks' is not a list with more than 3 elements")
else:
    print("fail: 'marks' key not found")
```

## question 3, check if dict is empty

```python
#using not operator
dict = {}
if not dict: #not returns True , else then False
    print("Empty")
else:
    print("Not empty")

```

```python
#by len()
dict = {}
if len(dict)==0:
    print("empty")
else:
    print("not empty")
```

## question 4, comparing two keys 

```python
marks = {"math":78,"science":80}
if marks["math"] > marks["science"]:
    print(f"math has greater value",marks["math"])
elif marks["math"] < marks["science"]:
    print(f"science has the greater value",marks["science"])
else:
    print("both are equal")
```

```python
#using max() function
marks = {"math":78,"science":80}
higher = max(marks,key=marks.get) # finds the maxmium value 
print(f"{higher} has the greatest value: {marks[higher]}")
```

## question no 8, checking count of keys

```python
counter = { "name":"yogs", "age":26,"pick_no":4,"marks":[1,2,3,4]}
count = sum(1 for value in counter.values() if isinstance(value, int))
print("Number of keys with integers values:",count)
```

```python
data = {"name":"yogs", "age":26,"pick_no":4,"marks":[1,2,3,4]}
count = 0
for key in data:
    if isinstance(data[key], int):
        count += 1
    else:
        pass 

print("Number of keys with integer values:", count)
```

## question 10, checking nested dict key

```python
data = {
    "user": {
        "admin": True,
        "name": "Alice"
    }
}

if "user" in data:
    if "admin" in data["user"]:
        if data["user"]["admin"] is True:
            print("Admin is True")
        else:
            print("Admin exists but is not True")
    else:
        print("Key 'admin' not found in 'user'")
```
