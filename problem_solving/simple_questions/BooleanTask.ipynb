```python
#question 3, 
def string_indicates(input_string):
    words = input_string.split()
    string_position = {}

    for index, word in enumerate(words):
        if word not in string_position:
            string_position[word] = []
            string_position[word].append(index)
    return string_position

text = ("my name is khan")
result = string_indicates(text)
print(result)
```

```python
def string_indices(input_string):
    words = input_string.split()
    strings_positions = {}

    for index, word in enumerate(words):
        if word not in strings_positions:
            strings_positions[word] = []
        strings_positions[word].append(index)
    return strings_positions
text = ("this is a test this is only a test")
result = string_indices(text)
print(result)
```
