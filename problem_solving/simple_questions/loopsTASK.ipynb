## multiplication from 2- 10, using for loop

```python
for i in range(2,11):
    print(f"Multiply of {i}:")
    for n in range(1,11):
        print(f"{i} * {n} = {i * n}")
    print()
```

## question no 3, multiplication of given nnumber n

```python
## using for loop
n = int(input("Enter the number:"))
print(f"multiplication table of {n}")
for i in range(1,11):
    print(f"{n} * {i} = {n * i}")
print()
```

```python
## using while loop
n = int(input("Enter the number:"))
print(f"multiplication of {n}:")
i = 1
while i <= 10:
    print(f"{n} * {i} = {n*i}")
    i += 1
```

## question no 4,  calculate the sum of no's from 1 - 50

```python
## using for loop
sum_no = 0
for i in range(1,51):
    sum_no += i
print("sum is",sum_no)
```

```python
## using while loop
sum_no = 0
i = 1
while i <= 50:
    sum_no += i
    i += 1
print("sum number is ",sum_no)
```
