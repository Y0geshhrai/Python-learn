## question 1, check if a string is a palindrome

```python
#using define function and asking user
def palindrome(s):
    if s == s[::-1]:
        print(f"{s} is a palindrome.")
    else:
        print(f"{s} is not a palindrome.")
user_input = input("enter a string")
palindrome(user_input)
```

## question 2, verify a sting contains only alphabets

```python
#asking user 
text = input("Enter a string: ")

if text.isalpha():
    print("The string contains only alphabets.")
else:
    print("The string contains non-alphabet characters.")
```

## question 8, count no. 


```python
vowles = ["a", "e","i","o","u"]
strings = input("enter the stings")
counter = 0
for character in strings:
    if character in vowles:
        counter = counter + 1
if counter>3:
    print(f"{strings} is a vowel rich")
else:
    print(f"{strings} is not a vowles")
```

## question 10,  check password length and no spaces

```python
def valid_password(password):
    if len(password) >= 8 and " " not in password:
        return True
    else:
        return False
password = input("enter your password: ")
if valid_password(password):
    print("your password is valid ")
else:
    print("invalid password")
```
